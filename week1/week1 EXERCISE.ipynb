```python
# imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
```

```python
# constants

# MODEL_GPT = 'gpt-4o-mini'
# MODEL_LLAMA = 'llama3.2'
MODEL_QWEN = 'qwen2.5:2b'
MODEL_DEEPSEEK = 'deepseek-v4-flash'
```

```python
# set up environment
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
deepseek_api_key = os.getenv('DEEPSEEK_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")

if deepseek_api_key:
    print(f"DeepSeek API Key exists and begins {deepseek_api_key[:8]}")
else:
    print("DeepSeek API Key not set")
    
MODEL = 'gpt-5-nano'
openai = OpenAI(
    api_key=os.getenv('DEEPSEEK_API_KEY'),
        base_url="https://api.deepseek.com"
)
```

```python
# here is the question; type over this to ask something new

question = """
Recommend me a gym schedule for 3-4 weeks for beginner. I want to build muscle and lose fat. I can go to the gym 4 times a week. Please provide a detailed schedule with exercises, sets, reps, and rest periods.
"""
```

```python
# messages
system_prompt = "You are a professional fitneess coach that helps people loose weight and get in shape."
user_prompt = "Please give a detailed explanation to the following question: " + question
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

```python
# Get stream_deepseek_answer to answer, with streaming

def stream_deepseek_answer(question):
    # Get DeepSeek to answer, with streaming
    stream = openai.chat.completions.create(model=MODEL_DEEPSEEK, messages=messages, stream=True)

    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ""
        update_display(Markdown(response), display_id=display_handle.display_id)

stream_deepseek_answer(question)
```
